Partikulat PM10 (debu berukuran &le; 10 mikrometer) berasal dari kendaraan bermotor, industri, dan konstruksi, dan dapat masuk ke saluran pernapasan sehingga memicu ISPA dan penyakit paru.   <h3 style="margin: 0; font-size: 16px; color: #FFFFFF;">1. IMPORT LIBRARY</h3>
  <p style="margin: 4px 0 0 0; font-size: 12px; color: #E0F7FA;">Memuat pustaka komputasi numerik, manipulasi data, visualisasi grafik, dan kerangka kerja PyTorch.</p>
</div>

```python
#202432022_BarruNaufalGhiyatsPranoto
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import torch
import torch.nn as nn
import torch.optim as optim
from torch.utils.data import TensorDataset, DataLoader
from sklearn.preprocessing import MinMaxScaler
from sklearn.metrics import mean_absolute_error, mean_squared_error

# Kunci seed agar hasil pelatihan bisa diulang (reproducible)
SEED = 42
np.random.seed(SEED)
torch.manual_seed(SEED)

# Gunakan GPU bila tersedia, jika tidak gunakan CPU
device = torch.device('cuda' if torch.cuda.is_available() else 'cpu')
print('Versi PyTorch :', torch.__version__)
print('Device        :', device)
```

<div style="background-color: #114F6A; color: #FFFFFF; padding: 10px 18px; border-radius: 8px; border-left: 8px solid #1FA4B1; font-family: 'Segoe UI', sans-serif; margin-top: 16px;">
  <h3 style="margin: 0; font-size: 16px; color: #FFFFFF;">2. PRA-PEMROSESAN DATA (PREPROCESSING)</h3>
  <p style="margin: 4px 0 0 0; font-size: 12px; color: #E0F7FA;">Memuat dataset, membersihkan data, melakukan normalisasi / tokenisasi sekuens, serta membagi data latih dan data uji.</p>
</div>

```python
#202432022_BarruNaufalGhiyatsPranoto
# 1. Muat dataset ISPU Stasiun DKI1 (Bunderan HI) dari Kaggle
FILE_PATH = 'ispu_dki1.csv'
TARGET = 'pm10'
df = pd.read_csv(FILE_PATH)
df.columns = df.columns.str.strip().str.lower()
print('Ukuran data mentah  :', df.shape)
print('Nama stasiun        :', df['stasiun'].unique().tolist())
display(df.head())

# 2. Pembersihan data & pra-pemrosesan (Forecasting)
df['tanggal'] = pd.to_datetime(df['tanggal'])
df[TARGET] = pd.to_numeric(df[TARGET], errors='coerce')
df = df.sort_values('tanggal').drop_duplicates(subset='tanggal')

# Tahun 2022 kosong total pada file stasiun, sehingga dipakai periode kontinu 2010-2021
series = df.set_index('tanggal')[TARGET].asfreq('D').loc['2010-01-01':'2021-12-31']
print('Rentang tanggal     :', series.index.min().date(), 's.d.', series.index.max().date())
print('Jumlah hari         :', len(series))
print('Nilai kosong (awal) :', series.isna().sum())
series = series.interpolate(method='time', limit_direction='both')
print('Nilai kosong (akhir):', series.isna().sum())
print(series.describe().round(2))

# Normalisasi Min-Max (scaler di-fit hanya pada data latih agar tidak terjadi kebocoran data)
values = series.values.reshape(-1, 1).astype(np.float32)
train_size = int(len(values) * 0.8)
scaler = MinMaxScaler(feature_range=(0, 1))
scaler.fit(values[:train_size])
scaled = scaler.transform(values).astype(np.float32)

# Sliding window: 30 hari sebelumnya -> prediksi 1 hari berikutnya
WINDOW = 30
def create_sequences(data, window):
    X, y = [], []
    for i in range(len(data) - window):
        X.append(data[i:i + window])
        y.append(data[i + window])
    return np.array(X), np.array(y)

X_train_np, y_train_np = create_sequences(scaled[:train_size], WINDOW)
X_test_np, y_test_np = create_sequences(scaled[train_size - WINDOW:], WINDOW)

# 3. Konversi data menjadi PyTorch Tensor (X_train, y_train, X_test, y_test)
X_train = torch.tensor(X_train_np, dtype=torch.float32)
y_train = torch.tensor(y_train_np, dtype=torch.float32)
X_test = torch.tensor(X_test_np, dtype=torch.float32)
y_test = torch.tensor(y_test_np, dtype=torch.float32)
print('X_train:', tuple(X_train.shape), '- y_train:', tuple(y_train.shape))
print('X_test :', tuple(X_test.shape), '- y_test :', tuple(y_test.shape))

# Plot deret waktu hasil pembersihan beserta batas data latih / data uji
plt.figure(figsize=(12, 4))
plt.plot(series.index, series.values, color='#114F6A', linewidth=1, label='ISPU PM10 harian')
plt.axvline(series.index[train_size], color='#1FA4B1', linestyle='--', linewidth=2, label='Batas data latih / uji')
plt.title('ISPU PM10 Harian Stasiun DKI1 Bunderan HI (2010-2021) Setelah Pembersihan')
plt.xlabel('Tanggal')
plt.ylabel('ISPU PM10')
plt.grid(True, linestyle='--', alpha=0.5)
plt.legend()
plt.show()
```

<div style="background-color: #114F6A; color: #FFFFFF; padding: 10px 18px; border-radius: 8px; border-left: 8px solid #1FA4B1; font-family: 'Segoe UI', sans-serif; margin-top: 16px;">
  <h3 style="margin: 0; font-size: 16px; color: #FFFFFF;">3. PEMODELAN ARSITEKTUR JARINGAN (RNN / LSTM)</h3>
  <p style="margin: 4px 0 0 0; font-size: 12px; color: #E0F7FA;">Mendefinisikan kelas arsitektur model sekuensial menggunakan modul torch.nn sesuai algoritma yang dipilih.</p>
</div>

```python
#202432022_BarruNaufalGhiyatsPranoto
class SequenceModel(nn.Module):
    def __init__(self, input_size=1, hidden_size=64, num_layers=2, output_size=1, dropout=0.2):
        super(SequenceModel, self).__init__()
        # Definisikan layer arsitektur: LSTM bertumpuk (stacked) + Linear untuk regresi
        self.hidden_size = hidden_size
        self.num_layers = num_layers
        self.lstm = nn.LSTM(input_size=input_size, hidden_size=hidden_size,
                            num_layers=num_layers, batch_first=True, dropout=dropout)
        self.fc = nn.Linear(hidden_size, output_size)

    def forward(self, x):
        # Hidden state (h0) dan cell state (c0) awal bernilai nol
        h0 = torch.zeros(self.num_layers, x.size(0), self.hidden_size, device=x.device)
        c0 = torch.zeros(self.num_layers, x.size(0), self.hidden_size, device=x.device)
        out, (hn, cn) = self.lstm(x, (h0, c0))
        # Ambil output langkah waktu terakhir, lalu petakan menjadi 1 nilai prediksi
        out = self.fc(out[:, -1, :])
        return out
```

<div style="background-color: #114F6A; color: #FFFFFF; padding: 10px 18px; border-radius: 8px; border-left: 8px solid #1FA4B1; font-family: 'Segoe UI', sans-serif; margin-top: 16px;">
  <h3 style="margin: 0; font-size: 16px; color: #FFFFFF;">4. INISIASI HYPERPARAMETER & PROSES PELATIHAN (TRAINING)</h3>
  <p style="margin: 4px 0 0 0; font-size: 12px; color: #E0F7FA;">Menentukan konfigurasi hyperparameter (learning rate, epochs, loss function, optimizer) dan menjalankan siklus pelatihan model.</p>
</div>

```python
#202432022_BarruNaufalGhiyatsPranoto
# 1. Inisiasi hyperparameter (epochs, learning_rate, hidden_size, dll.)
epochs = 50
learning_rate = 0.001
hidden_size = 64
num_layers = 2
dropout = 0.2
batch_size = 32
train_loader = DataLoader(TensorDataset(X_train, y_train), batch_size=batch_size, shuffle=True)

# 2. Inisiasi objek model, fungsi Loss (MSELoss), dan Optimizer (Adam)
model = SequenceModel(input_size=1, hidden_size=hidden_size, num_layers=num_layers,
                      output_size=1, dropout=dropout).to(device)
criterion = nn.MSELoss()
optimizer = optim.Adam(model.parameters(), lr=learning_rate)
print(model)

# 3. Simpan riwayat nilai loss pada list untuk divisualisasikan
train_losses = []
test_losses = []

# 4. Jalankan proses perulangan training (Training Loop)
for epoch in range(epochs):
    model.train()
    running_loss = 0.0
    for X_batch, y_batch in train_loader:
        X_batch, y_batch = X_batch.to(device), y_batch.to(device)
        optimizer.zero_grad()
        outputs = model(X_batch)
        loss = criterion(outputs, y_batch)
        loss.backward()
        optimizer.step()
        running_loss += loss.item() * X_batch.size(0)
    train_losses.append(running_loss / len(train_loader.dataset))

    model.eval()
    with torch.no_grad():
        test_loss = criterion(model(X_test.to(device)), y_test.to(device)).item()
    test_losses.append(test_loss)

    if (epoch + 1) % 5 == 0 or epoch == 0:
        print(f'Epoch [{epoch + 1:2d}/{epochs}] - Train Loss: {train_losses[-1]:.6f} - Test Loss: {test_loss:.6f}')
```

<div style="background-color: #114F6A; color: #FFFFFF; padding: 10px 18px; border-radius: 8px; border-left: 8px solid #1FA4B1; font-family: 'Segoe UI', sans-serif; margin-top: 16px;">
  <h3 style="margin: 0; font-size: 16px; color: #FFFFFF;">5. VISUALISASI LOSS TRAINING & PERBANDINGAN PREDIKSI VS AKTUAL</h3>
  <p style="margin: 4px 0 0 0; font-size: 12px; color: #E0F7FA;">Menampilkan grafik konvergensi nilai Loss terhadap Epoch dan grafik evaluasi perbandingan hasil prediksi model terhadap data aktual.</p>
</div>

```python
#202432022_BarruNaufalGhiyatsPranoto
# 1. Plot Grafik Penurunan Loss Training terhadap Epoch
plt.figure(figsize=(10, 4))
plt.plot(range(1, epochs + 1), train_losses, color='#114F6A', linewidth=2, label='Training Loss')
plt.plot(range(1, epochs + 1), test_losses, color='#1FA4B1', linewidth=2, linestyle='--', label='Test Loss')
plt.title('Grafik Penurunan Loss Selama Pelatihan')
plt.xlabel('Epoch')
plt.ylabel('Loss (MSE)')
plt.grid(True, linestyle='--', alpha=0.5)
plt.legend()
plt.show()

# Prediksi data uji, lalu kembalikan ke skala ISPU asli (inverse transform)
model.eval()
with torch.no_grad():
    y_pred_scaled = model(X_test.to(device)).cpu().numpy()
y_pred = scaler.inverse_transform(y_pred_scaled).flatten()
y_true = scaler.inverse_transform(y_test.numpy()).flatten()
test_dates = series.index[train_size:]

# Metrik evaluasi + pembanding naif (prediksi besok = nilai hari ini)
mae = mean_absolute_error(y_true, y_pred)
rmse = np.sqrt(mean_squared_error(y_true, y_pred))
mape = np.mean(np.abs((y_true - y_pred) / y_true)) * 100
naive_pred = values[train_size - 1:-1].flatten()
mae_naive = mean_absolute_error(y_true, naive_pred)
print(f'MAE  LSTM  : {mae:.2f}')
print(f'RMSE LSTM  : {rmse:.2f}')
print(f'MAPE LSTM  : {mape:.2f}%')
print(f'MAE  Naif  : {mae_naive:.2f}  (pembanding: prediksi besok = nilai hari ini)')

# 2. Plot Grafik Perbandingan Data Aktual vs Hasil Prediksi Model
plt.figure(figsize=(12, 5))
plt.plot(test_dates, y_true, color='#114F6A', linewidth=1.5, label='Data Aktual')
plt.plot(test_dates, y_pred, color='#1FA4B1', linewidth=1.5, label='Prediksi LSTM')
plt.title('Perbandingan Data Aktual vs Prediksi ISPU PM10 (Data Uji)')
plt.xlabel('Tanggal')
plt.ylabel('ISPU PM10')
plt.grid(True, linestyle='--', alpha=0.5)
plt.legend()
plt.show()

# Perbesar 90 hari terakhir agar pola prediksi terlihat lebih jelas
plt.figure(figsize=(12, 5))
plt.plot(test_dates[-90:], y_true[-90:], color='#114F6A', linewidth=2, label='Data Aktual')
plt.plot(test_dates[-90:], y_pred[-90:], color='#1FA4B1', linewidth=2, linestyle='--', label='Prediksi LSTM')
plt.title('Zoom 90 Hari Terakhir: Data Aktual vs Prediksi ISPU PM10')
plt.xlabel('Tanggal')
plt.ylabel('ISPU PM10')
plt.grid(True, linestyle='--', alpha=0.5)
plt.legend()
plt.show()
```
